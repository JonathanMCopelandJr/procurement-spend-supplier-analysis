# Procurement Spend & Supplier Performance Analysis
**Synthetic portfolio case study** | Python • SQL • KPI reporting • supplier performance

## Objective
Analyze purchase-order data to identify spend concentration, supplier delivery risk, and priorities for procurement action. All data is synthetic and created for portfolio demonstration only.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
sns.set_theme(style='whitegrid')
ROOT = Path('..') if Path('../data/synthetic_procurement_data.csv').exists() else Path('.')
df = pd.read_csv(ROOT / 'data' / 'synthetic_procurement_data.csv', parse_dates=['order_date','required_delivery_date','actual_delivery_date'])
df.head()
```

```python
df['on_time_flag'] = (df['delivery_status'] == 'On Time').astype(int)
summary = df.groupby('supplier').agg(total_spend=('total_spend','sum'), purchase_orders=('po_number','count'), on_time_delivery_pct=('on_time_flag', lambda x: x.mean()*100), avg_days_late=('days_late','mean'), avg_quality_score=('quality_score','mean')).reset_index().sort_values('total_spend', ascending=False)
summary.round(2)
```

```python
plot = summary.head(8).sort_values('total_spend')
plt.figure(figsize=(11,6))
plt.barh(plot['supplier'], plot['total_spend']/1000, color='#00A6A6')
plt.title('Top Supplier Spend ($000s)', fontweight='bold')
plt.xlabel('Spend ($000s)')
plt.tight_layout()
plt.show()
```

## Decision lens
Prioritize supplier-development plans or alternate-source evaluation for suppliers with below-target on-time delivery, repeated late orders, or weak quality scores—especially where spend exposure is high.
